## Solution

In this section, we will build a reusable web scraping function called `scrape_books`.

The function will:

1. Visit all catalogue pages from **Books to Scrape**.
2. Extract the link of each book.
3. Visit each individual book page to collect detailed information.
4. Filter the results by:
   - minimum rating
   - maximum price
5. Return the final result as a pandas DataFrame.

The expected output columns are:

- `UPC`
- `Title`
- `Price (£)`
- `Rating`
- `Genre`
- `Availability`
- `Description`


```python
# Import the required libraries

import requests
import pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import time 
import re
```

```python
def scrape_books(min_rating, max_price):
    """
    Scrape book data from Books to Scrape and return a filtered pandas DataFrame.

    Parameters
    ----------
    min_rating : int or float
        Minimum rating required for a book to be included.
    max_price : int or float
        Maximum book price in pounds.

    Returns
    -------
    pandas.DataFrame
        DataFrame with UPC, Title, Price (£), Rating, Genre, Availability and Description.
    """

    base_url = "http://books.toscrape.com/"
    catalogue_url = urljoin(base_url, "catalogue/")

    rating_map = {
        "One": 1,
        "Two": 2,
        "Three": 3,
        "Four": 4,
        "Five": 5
    }

    books_data = []
    page_number = 1

    while True:
        page_url = urljoin(catalogue_url, f"page-{page_number}.html")
        response = requests.get(page_url)

        if response.status_code != 200:
            break

        soup = BeautifulSoup(response.text, "html.parser")
        books = soup.select("article.product_pod")

        if not books:
            break

        for book in books:
            rating_class = book.select_one("p.star-rating").get("class", [])
            rating_word = [item for item in rating_class if item != "star-rating"][0]
            rating = rating_map.get(rating_word)

            price_text = book.select_one("p.price_color").get_text(strip=True)
            price = float(re.sub(r"[^\d.]", "", price_text))

            if rating < min_rating or price > max_price:
                continue

            relative_book_url = book.select_one("h3 a").get("href")
            book_url = urljoin(page_url, relative_book_url)

            book_response = requests.get(book_url)

            if book_response.status_code != 200:
                continue

            book_soup = BeautifulSoup(book_response.text, "html.parser")

            title = book_soup.select_one("div.product_main h1").get_text(strip=True)

            breadcrumb_items = book_soup.select("ul.breadcrumb li a")
            genre = breadcrumb_items[-1].get_text(strip=True) if breadcrumb_items else None

            availability = book_soup.select_one("p.instock.availability").get_text(strip=True)

            description_heading = book_soup.find("div", id="product_description")
            if description_heading:
                description = description_heading.find_next_sibling("p").get_text(strip=True)
            else:
                description = None

            product_info_rows = book_soup.select("table.table.table-striped tr")
            product_info = {}

            for row in product_info_rows:
                key = row.select_one("th").get_text(strip=True)
                value = row.select_one("td").get_text(strip=True)
                product_info[key] = value

            upc = product_info.get("UPC")

            books_data.append({
                "UPC": upc,
                "Title": title,
                "Price (£)": price,
                "Rating": rating,
                "Genre": genre,
                "Availability": availability,
                "Description": description
            })

            time.sleep(0.1)

        page_number += 1

    expected_columns = [
        "UPC",
        "Title",
        "Price (£)",
        "Rating",
        "Genre",
        "Availability",
        "Description"
    ]

    df = pd.DataFrame(books_data)
    return df.reindex(columns=expected_columns)
```

```python
books_df = scrape_books(min_rating=4, max_price=20)
books_df.head()
```

```python
# Check the shape of the resulting DataFrame

books_df.shape
```

```python
# Quick summary by genre

books_df.groupby("Genre").agg(
    number_of_books=("Title", "count"),
    average_price=("Price (£)", "mean"),
    average_rating=("Rating", "mean")
).sort_values("number_of_books", ascending=False)
```

```python
# Optional: export the scraped data to a CSV file

books_df.to_csv("books_to_scrape_filtered.csv", index=False)
```
