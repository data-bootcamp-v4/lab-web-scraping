# LAB | Web Scraping

## Books to Scrape

En este lab vamos a construir un pequeño scraper para extraer información de libros desde [Books to Scrape](https://books.toscrape.com/).

El objetivo es practicar cómo usar **requests**, **BeautifulSoup** y **pandas** para obtener datos desde una página web, filtrarlos según ciertos criterios y guardarlos en un DataFrame listo para analizar.

### Objetivo del lab

Crear una función que scrapee libros según:

- una **calificación mínima** (`min_rating`)
- un **precio máximo** (`max_price`)

Y que devuelva un DataFrame con información relevante de cada libro.

## 1. Importar librerías

Primero importamos las librerías que vamos a necesitar:

- `requests`: para hacer peticiones HTTP a la web.
- `BeautifulSoup`: para leer y navegar el HTML.
- `pandas`: para organizar los datos en formato tabla.
- `time`: para añadir una pequeña pausa entre páginas y evitar hacer peticiones demasiado agresivas.

````python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
````

## 2. Conocer la página objetivo

La web que vamos a scrapear es:

`https://books.toscrape.com/`

Esta página fue creada precisamente para practicar web scraping, así que es perfecta para este ejercicio.

Cada libro tiene información como:

- título
- precio
- disponibilidad
- rating
- enlace al detalle del libro

Primero vamos a hacer una petición simple a la página principal para verificar que podemos acceder correctamente.

````python
base_url = "https://books.toscrape.com/"

response = requests.get(base_url)

print(response.status_code)
````

Si el resultado es `200`, significa que la petición fue exitosa.

Un código `200` es básicamente la web diciendo: *sí, pasa, te estaba esperando con cafecito y HTML*.

## 3. Parsear el HTML con BeautifulSoup

Ahora convertimos el contenido HTML en un objeto que BeautifulSoup pueda leer y recorrer.

````python
soup = BeautifulSoup(response.content, "html.parser")

# Visualizamos una pequeña parte del HTML
print(soup.prettify()[:1000])
````

## 4. Encontrar los libros en una página

Inspeccionando el HTML de la página, vemos que cada libro está dentro de un elemento:

```html
<article class="product_pod">
```

Por eso podemos usar `find_all()` para capturar todos los libros de la página actual.

````python
books = soup.find_all("article", class_="product_pod")

len(books)
````

La página muestra 20 libros por página.

Ahora vamos a probar cómo extraer la información del primer libro.

````python
first_book = books[0]

first_book
````

## 5. Extraer título, precio, rating y disponibilidad

Vamos a extraer los campos principales de un libro individual.

En esta web, el rating aparece como una clase CSS, por ejemplo:

```html
<p class="star-rating Three">
```

Por eso necesitaremos convertir palabras como `One`, `Two`, `Three`, etc. en números.

````python
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}
````

````python
title = first_book.h3.a["title"]
price = first_book.find("p", class_="price_color").text
availability = first_book.find("p", class_="instock availability").text.strip()
rating_text = first_book.find("p", class_="star-rating")["class"][1]
rating = rating_map[rating_text]
relative_link = first_book.h3.a["href"]
book_url = base_url + relative_link

print("Title:", title)
print("Price:", price)
print("Availability:", availability)
print("Rating:", rating)
print("URL:", book_url)
````

## 6. Limpiar el precio

El precio viene como texto, por ejemplo `£51.77`.

Para poder compararlo con un precio máximo, necesitamos convertirlo a número decimal.

````python
price_text = first_book.find("p", class_="price_color").text
price_float = float(price_text.replace("£", ""))

price_float
````

## 7. Crear una función para scrapear una página

Ahora que ya sabemos extraer información de un libro, vamos a crear una función que haga esto para todos los libros de una página.

````python
def scrape_books_from_page(page_url):
    """
    Scrapes book information from a single Books to Scrape page.
    
    Parameters:
    page_url (str): URL of the page to scrape.
    
    Returns:
    list: A list of dictionaries containing book data.
    """
    response = requests.get(page_url)
    
    if response.status_code != 200:
        print(f"Failed to retrieve page: {page_url}")
        return []
    
    soup = BeautifulSoup(response.content, "html.parser")
    books = soup.find_all("article", class_="product_pod")
    
    books_data = []
    
    for book in books:
        title = book.h3.a["title"]
        price_text = book.find("p", class_="price_color").text
        price = float(price_text.replace("£", ""))
        availability = book.find("p", class_="instock availability").text.strip()
        
        rating_text = book.find("p", class_="star-rating")["class"][1]
        rating = rating_map[rating_text]
        
        relative_link = book.h3.a["href"]
        
        # If the link is from the homepage, it does not include "catalogue/"
        if "catalogue/" not in relative_link:
            book_url = base_url + "catalogue/" + relative_link
        else:
            book_url = base_url + relative_link
        
        books_data.append({
            "title": title,
            "price": price,
            "rating": rating,
            "availability": availability,
            "url": book_url
        })
    
    return books_data
````

Probamos la función con la primera página.

````python
page_1_books = scrape_books_from_page(base_url)

pd.DataFrame(page_1_books).head()
````

## 8. Scraping de múltiples páginas

La web tiene varias páginas. La estructura de URLs después de la primera página es:

```text
https://books.toscrape.com/catalogue/page-2.html
https://books.toscrape.com/catalogue/page-3.html
...
```

Vamos a recorrer todas las páginas disponibles. Books to Scrape tiene 50 páginas en total.

````python
all_books = []

# Page 1 has a different URL
all_books.extend(scrape_books_from_page(base_url))

# Pages 2 to 50 follow a predictable pattern
for page_num in range(2, 51):
    page_url = f"https://books.toscrape.com/catalogue/page-{page_num}.html"
    books_on_page = scrape_books_from_page(page_url)
    all_books.extend(books_on_page)
    time.sleep(0.5)  # polite scraping

len(all_books)
````

## 9. Crear el DataFrame

Ahora convertimos la lista de diccionarios en un DataFrame de pandas.

````python
books_df = pd.DataFrame(all_books)

books_df.head()
````

````python
books_df.info()
````

## 10. Filtrar por rating mínimo y precio máximo

El lab nos pide extraer datos basados en criterios específicos:

- rating mínimo
- precio máximo

Vamos a crear una función que reciba esos criterios y devuelva solamente los libros que cumplen las condiciones.

````python
def filter_books(df, min_rating=4, max_price=30):
    """
    Filters books according to minimum rating and maximum price.
    
    Parameters:
    df (DataFrame): DataFrame containing book data.
    min_rating (int): Minimum rating required.
    max_price (float): Maximum price allowed.
    
    Returns:
    DataFrame: Filtered DataFrame.
    """
    filtered_df = df[(df["rating"] >= min_rating) & (df["price"] <= max_price)].copy()
    filtered_df = filtered_df.sort_values(by=["rating", "price"], ascending=[False, True])
    
    return filtered_df
````

Probamos la función con:

- rating mínimo: 4
- precio máximo: £30

````python
filtered_books = filter_books(books_df, min_rating=4, max_price=30)

filtered_books.head(10)
````

````python
filtered_books.shape
````

## 11. Función final del lab

Ahora vamos a unir todo en una función final que:

1. Scrapea todas las páginas.
2. Crea un DataFrame.
3. Filtra según `min_rating` y `max_price`.
4. Devuelve el DataFrame final.

Esta sería la función más útil para entregar como solución del lab.

````python
def scrape_books(min_rating=4, max_price=30):
    """
    Scrapes Books to Scrape and returns books filtered by minimum rating and maximum price.
    
    Parameters:
    min_rating (int): Minimum rating required. Values should be between 1 and 5.
    max_price (float): Maximum price allowed.
    
    Returns:
    DataFrame: Filtered books DataFrame.
    """
    all_books = []
    
    # First page
    all_books.extend(scrape_books_from_page(base_url))
    
    # Remaining pages
    for page_num in range(2, 51):
        page_url = f"https://books.toscrape.com/catalogue/page-{page_num}.html"
        all_books.extend(scrape_books_from_page(page_url))
        time.sleep(0.5)
    
    df = pd.DataFrame(all_books)
    filtered_df = filter_books(df, min_rating=min_rating, max_price=max_price)
    
    return filtered_df
````

````python
final_books_df = scrape_books(min_rating=4, max_price=30)

final_books_df.head(10)
````

## 12. Análisis rápido

Una vez tenemos los datos filtrados, podemos hacer algunas preguntas simples:

- ¿Cuántos libros cumplen los criterios?
- ¿Cuál es el precio promedio?
- ¿Cuántos libros hay por rating?

````python
print("Number of books:", final_books_df.shape[0])
print("Average price:", round(final_books_df["price"].mean(), 2))
print("Minimum price:", final_books_df["price"].min())
print("Maximum price:", final_books_df["price"].max())
````

````python
final_books_df["rating"].value_counts().sort_index()
````

## 13. Guardar los resultados en CSV

Finalmente, guardamos el DataFrame filtrado en un archivo CSV.

````python
final_books_df.to_csv("filtered_books.csv", index=False)

print("CSV file saved successfully!")
````
